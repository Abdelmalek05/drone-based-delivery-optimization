# Greedy Algorithm - Drone Delivery Route Optimization

This notebook now mirrors the exact code from `greedy.py` and explains each block of the implementation and the result interpretation.

```python
"""Greedy Search Algorithm for Drone Delivery Route Optimization."""

from __future__ import annotations

import random
import sys
from pathlib import Path
from time import perf_counter
from typing import Dict, List, Tuple

if __package__ is None or __package__ == "":
    sys.path.append(str(Path(__file__).resolve().parent))

from solver_base import Problem, greedy_solve
```

### Imports and module setup

This cell contains the exact import and path-handling logic from `greedy.py`.

- `random` is used to seed deterministic runs.
- `perf_counter` measures runtime.
- `Problem` loads the optimization instance.
- `greedy_solve` performs the actual greedy route construction.
- The `sys.path` block makes the module runnable both as a script and as a notebook-style import.

```python
def solution_summary(problem: Problem, solution: List[Tuple[int, List[int]]]) -> Dict[str, float]:
    """Compute comparable metrics for a solution."""
    fitness, energy, violations, unserved = problem.evaluate(solution)
    served = problem.n - len(unserved)
    route_count = sum(1 for _, route in solution if route)
    return {
        "fitness": float(fitness),
        "energy_kwh": float(energy),
        "violations": int(violations),
        "unserved": int(len(unserved)),
        "served": int(served),
        "routes": int(route_count),
    }
```

### `solution_summary`

This helper is responsible for turning a raw solution into comparable metrics.

It reports:
- fitness
- energy in kWh
- constraint violations
- unserved customers
- served customers
- number of routes used

This is the function that makes the greedy notebook comparable to the genetic notebook.

```python
def run_greedy(seed: int = 42) -> Dict[str, object]:
    """Solve the problem with the greedy heuristic and return all artifacts."""
    random.seed(seed)
    problem = Problem()

    start = perf_counter()
    solution = greedy_solve(problem)
    elapsed = perf_counter() - start

    metrics = solution_summary(problem, solution)
    metrics["runtime_sec"] = elapsed

    return {
        "problem": problem,
        "solution": solution,
        "metrics": metrics,
    }
```

### `run_greedy`

This cell reproduces the full greedy execution path exactly as in `greedy.py`.

Interpretation:
- the random seed makes the run reproducible
- the `Problem` instance loads the optimization model
- `greedy_solve(problem)` computes the route solution
- runtime is measured around the solver call
- the returned dictionary keeps the problem, solution, and metrics together

```python
def print_report(problem: Problem, solution: List[Tuple[int, List[int]]], metrics: Dict[str, float]) -> None:
    print("=" * 70)
    print("GREEDY ALGORITHM - Drone Delivery Route Optimization")
    print("=" * 70)
    print(problem.format(solution))
    print("-" * 70)
    print(f"Total Energy:          {metrics['energy_kwh']:.3f} kWh")
    print(f"Constraint Violations: {metrics['violations']}")
    print(f"Unserved Customers:    {metrics['unserved']}")
    print(f"Routes Used:           {metrics['routes']}")
    print(f"Runtime:               {metrics['runtime_sec']:.4f} s")
    if metrics["unserved"]:
        _, _, _, unserved = problem.evaluate(solution)
        print(f"  -> {[problem.customers[i]['name'] for i in sorted(unserved)]}")
    print("=" * 70)
```

### `print_report`

This cell prints the human-readable result summary exactly as the script does.

It is useful because it shows:
- the formatted route plan
- total energy
- constraint violations
- unserved customers
- routes used
- runtime

If any customers remain unserved, their names are listed explicitly.

```python
def main() -> Dict[str, object]:
    result = run_greedy()
    problem = result["problem"]
    solution = result["solution"]
    metrics = result["metrics"]
    print_report(problem, solution, metrics)
    return result


if __name__ == "__main__":
    main()
```
